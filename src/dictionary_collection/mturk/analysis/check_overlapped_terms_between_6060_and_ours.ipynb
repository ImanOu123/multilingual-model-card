```python
import pandas as pd
import numpy as np
langs = ['Arabic', 'Chinese', 'French', 'Japanese', 'Russian']

for lang in langs:
    df_6060 = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/src/dictionary_collection/mturk/analysis/annotation_results_6060/{lang}_validated.csv")
    
    word_6060 = set(df_6060['word'].tolist())
    intersection = list(word_6060)
    
    data0, data1, data2 = [], [], []
    for item in intersection:
        data0.append(item)
        data1.append(df_6060[df_6060['word'] == item]['validated_translation'].item())
        data2.append(df_6060[df_6060['word'] == item]['gold'].item())
    
    df = pd.DataFrame()
    df['word'] = data0
    df['annotation6060'] = data1
    df['label60'] = data2
    df_tmp = df[df['annotation6060'] != df['label60']]
    df_tmp.to_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/src/dictionary_collection/mturk/analysis/annotation_results_crawled/check_agreement_on_6060/{lang}.csv")
    
    print(lang)
    print("annotate6060 vs label6060", df[df['annotation6060'] == df['label60']].shape[0] / df.shape[0])
```

```python
import json
from tqdm import tqdm
from multilingualmc.methods.utils import DocProcessor, split_paragraph

tgt_langs = [
    "Chinese",
    "Arabic",
    "French",
    "Russian",
    "Japanese"
]
info_file = "info.json"
in_dir = "/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_ai_papers/text/"
json_info = json.load(open(info_file, 'r'))
paper_paths = [in_dir + paper for paper in json_info['test']['paper']]

paper_chunks = []
for paper_path in paper_paths:
    doc_store = DocProcessor(paper_path)
    doc = doc_store.doc
    
    for item in doc:
        if item['heading'] not in ['title', 'authors']:
            content_chunks = split_paragraph(item['content'])
            for chunk in content_chunks:
                
                info = {
                    "file": paper_path.split("/")[-1],
                    "source": "paper",
                    "heading": item['heading'],
                    "text": chunk,
                }
                paper_chunks.append(info)
```

```python
in_dir = "/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_model_cards/original/claude3_json/"
model_card_paths = [in_dir + model_card for model_card in json_info['test']['model_card']]

model_card_chunks = []
for model_card_path in model_card_paths:
    json_list = json.load(open(model_card_path, 'r'))
    for idx, item in enumerate(json_list):
        content_chunks = split_paragraph(item['answer'])
        for chunk in content_chunks:
            info = {
                "file": model_card_path.split("/")[-1],
                "source": "model_card",
                "heading": item['question'],
                "text": chunk,
            }
            model_card_chunks.append(info)
```

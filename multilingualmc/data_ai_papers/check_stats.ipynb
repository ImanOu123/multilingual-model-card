```python
import json
import os
def get_json_list(path):
    import json
    f = open(path, 'r')
    info = []
    for line in f.readlines():
        info.append(json.loads(line))
    return info
confs = [
    "aaai", 'acl', 'coling', 'conll', 'cvpr', 'eacl', 'eccv', 'emnlp', 'iccv', 'ijcai', 'lrec', 'naacl'
]
in_dir = "json/"
out_dir = "new_json/"
pdf_dir = "pdf/"
pdfs = os.listdir(pdf_dir)
data = []
for conf in confs:
    in_data = json.load(open(in_dir + conf + "_best_papers.json", 'r'))
    out_data = get_json_list(out_dir + conf + "_best_papers.jsonl")
    # if len(in_data) != len(out_data):
    #     print(conf)
    no_hash_cnt = 0
    has_pdf = 0
    cnt = len(out_data)
    for item in out_data:
        if item['file_hash'] is None:
            no_hash_cnt += 1
        else:
            if item['file_hash'] + '.pdf' in pdfs:
                has_pdf += 1
    info = {
        "conference": conf,
        "count": cnt,
        "has_pdf": has_pdf,
        "has_pdf_ratio": has_pdf / cnt,
        "has_hash": cnt - no_hash_cnt,
        "hash_hash_ratio": (cnt - no_hash_cnt) / cnt,
    }
    data.append(info)
```

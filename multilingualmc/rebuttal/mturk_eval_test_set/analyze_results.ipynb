```python
import pandas as pd

tgt_langs = [
    "Chinese",
    "Arabic",
    "French",
    "Japanese",
    "Russian",
]

res = []

for tgt_lang in tgt_langs:
    df = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/rebuttal/mturk_eval_test_set/results/annotation_{tgt_lang}.csv")
    
    # Calculate q1 mean and std
    q1_scores = sum(int(col.split('.')[1]) * df[col] for col in df.columns if 'q1' in col)
    q1_mean = q1_scores.mean()
    q1_std = q1_scores.std()

    # Calculate q2 and q3 mean and std
    q2_scores = df["Answer.Yes.q2"].astype(int)  # Yes = 1, No = 0
    q3_scores = df["Answer.Yes.q3"].astype(int)  # Yes = 1, No = 0

    q2_mean, q2_std = q2_scores.mean(), q2_scores.std()
    q3_mean, q3_std = q3_scores.mean(), q3_scores.std()

    # Output results
    results = {
        "Language": tgt_lang,
        "Q1": f"{q1_mean:.2f} ± {q1_std:.2f}",
        "Q2": f"{q2_mean:.2f} ± {q2_std:.2f}",
        "Q3": f"{q3_mean:.2f} ± {q3_std:.2f}",
    }
    res.append(results)

df = pd.DataFrame(res)
df
```

```python
import json
import pandas as pd
import os
from collections import defaultdict

dir_path = "/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_ai_papers/new_json/"

data = defaultdict(dict)
for file in os.listdir(dir_path):
    file_path = dir_path + file
    file_data = [json.loads(i) for i in open(file_path, 'r').readlines()]
    
    for item in file_data:
        data[item['file_hash']] = [int(item['year']), file.split("_best_papers.jsonl")[0]]
```

```python
from tqdm import tqdm
from collections import Counter
import re

data_terms = [json.loads(i) for i in open("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/growing_dict/terms.jsonl", 'r').readlines()]

terms_over_years = defaultdict(list)
paper_path = None
paper_terms_curr = []
for item in tqdm(data_terms):
    try:
        year = data[item['paper_path'].split(".json")[0]][0]
    except:
        continue
        
    paper_terms = [re.sub(r'\(.*?\)', '', i.lower()).strip() for i in item['processed_result']]
    if item['paper_path'] == paper_path:
        paper_terms_curr.extend(paper_terms)
    else:
        terms_over_years[year].extend(list(set(paper_terms_curr)))
        paper_terms_curr = paper_terms
        paper_path = item['paper_path']

terms_over_years[year].extend(list(set(paper_terms_curr)))

terms_over_years_counter = {}

for year, terms in terms_over_years.items():
    item_counts = Counter(terms)
    terms_over_years_counter[year] = dict(sorted(item_counts.items(), key=lambda x: x[1], reverse=True))
```

```python
terms = pd.read_csv("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/mturk/analysis/annotation_final/Arabic.csv")['English'].tolist()
terms = {term.lower(): term for term in terms}
```

```python
res = {}
for i in range(2000, 2024):
    cnt = 0
    res[i] = {}
    
    for term in terms_over_years_counter[i]:
        if cnt >= 10:
            break
        
        if term in terms and term not in ['', 'none']:
            cnt += 1
            res[i][terms[term]] = terms_over_years_counter[i][term]
            
```

```python
df = pd.DataFrame()
df['Year'] = [i for i in range(2000, 2024)]
df['Terms'] = [str(res[i]) for i in range(2000, 2024)]
print(df.to_markdown(index=False))
```

```python
# alternative way
res = {}
for i in range(2000, 2024):
    cnt = 0
    res[i] = []
    
    for term in terms_over_years_counter[i]:
        if cnt >= 10:
            break
        
        if term in terms and term not in ['', 'none']:
            cnt += 1
            res[i].append(terms[term])
            
df = pd.DataFrame()
df['Year'] = [i for i in range(2000, 2024)]
df['Terms'] = [", ".join(res[i]) for i in range(2000, 2024)]
print(df.to_markdown(index=False))
```

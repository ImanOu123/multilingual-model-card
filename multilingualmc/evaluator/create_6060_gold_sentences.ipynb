```python
import pandas as pd
import json
tgt_langs = [
    "Chinese",
    "Arabic",
    "French",
    "Japanese",
    "Russian",
]
gt_file = "/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/2/acl_6060/dev/text/txt/ACL.6060.dev.en-xx.en.txt"
gt_dict = {
    "English": [i for i in open(gt_file, 'r').readlines()],
    "Chinese": [i for i in open(gt_file.replace("en.txt", "zh.txt"), 'r').readlines()],
    "Arabic": [i for i in open(gt_file.replace("en.txt", "ar.txt"), 'r').readlines()],
    "French": [i for i in open(gt_file.replace("en.txt", "fr.txt"), 'r').readlines()],
    "Japanese": [i for i in open(gt_file.replace("en.txt", "ja.txt"), 'r').readlines()],
    "Russian": [i for i in open(gt_file.replace("en.txt", "ru.txt"), 'r').readlines()],
}


out_f = open("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/output/gold_dev_prompt_gpt4o.jsonl", 'a')
for i in range(len(gt_dict['English'])):
    info = {
        "text": gt_dict['English'][i],
    }
    for lang in tgt_langs:
        info[f"text_{lang}"] = gt_dict[lang][i]
    json.dump(info, out_f, ensure_ascii=False)
    out_f.write("\n")
    out_f.flush()
```

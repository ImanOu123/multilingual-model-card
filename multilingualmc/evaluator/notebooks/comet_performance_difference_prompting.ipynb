```python
import json
import pandas as pd
res_default_comet = json.load(open("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/eval/predictions_dev_seamless_comet.json", 'r'))
res_default = [json.loads(i) for i in open("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/output/predictions_dev_seamless.jsonl", 'r').readlines()]

res_prompt_comet = json.load(open("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/eval/predictions_dev_seamless_prompt_gpt4omini_comet.json", 'r'))
res_prompt = [json.loads(i) for i in open("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/output/predictions_dev_seamless_prompt_gpt4omini.jsonl", 'r').readlines()]
```

```python
df = pd.DataFrame()
tgt_langs = [
    "Chinese",
    "Arabic",
    "French",
    "Japanese",
    "Russian",
]

for lang in tgt_langs:
    df[f'text_English'] = [i['text'] for i in res_default]
    df[f"default_comet_{lang}"] = res_default_comet[f'{lang}_scores']
    df[f'default_text_{lang}'] = [i[f'text_{lang}'] for i in res_default]
    df[f"prompt_comet_{lang}"] = res_prompt_comet[f'{lang}_scores']
    df[f'prompt_text_{lang}'] = [i[f'text_{lang}'] for i in res_prompt]

df
```

```python
import matplotlib.pyplot as plt
import numpy as np
def draw_bar(correct_lst, lang):
    # Define the bin width
    bins = np.arange(-0.3, 0.3, 0.01).tolist()

    # Create the binplot (histogram)
    plt.figure(figsize=(8, 6))
    plt.hist(correct_lst, bins=bins, edgecolor='black', color='blue', alpha=0.6)

    # Add labels and title
    plt.xlabel('Value Range')
    plt.ylabel('Counts')
    plt.title(f'Binplot for {lang}')
    plt.legend()
    # Show the plot
    plt.show()
```

```python
for lang in tgt_langs:
    draw_bar((df[f'prompt_comet_{lang}'] - df[f'default_comet_{lang}']).tolist(), lang)
```

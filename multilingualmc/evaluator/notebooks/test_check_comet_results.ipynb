```python
import json
import pandas as pd
tgt_langs = [
    "Chinese",
    "Arabic",
    "French",
    "Japanese",
    "Russian",
]
```

```python
# models = ['gpt4omini', 'nllb', 'seamless', 'aya', 'aya_old']
models = ['aya_old']
exp_dict = {"default": "_comet", "prompt": "_prompt_gpt4omini_comet", "replace": "_hard_replace_comet"}
dataframe_dict = {}
for model in models:
    for exp_name, exp_suffix in exp_dict.items():
        column_name = f"{exp_name}_{model}"
        dataframe_dict[column_name] = []
        
        data = json.load(open(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/eval/gold_predictions_dev_{model}{exp_suffix}.json"))
        for lang in tgt_langs:
            dataframe_dict[column_name].append(data.get(f"{lang}_system_score", 0))  # Use `.get()` with default value `0` if lang not found

# Create the DataFrame
df = pd.DataFrame(dataframe_dict, index=tgt_langs)
df
```

```python
from collections import defaultdict
scores_dict = defaultdict(list)
for tgt_lang in tgt_langs:
    # print(f"{tgt_lang}_system_score", data[f"{tgt_lang}_system_score"])
    scores_dict[f"{tgt_lang}_system_score"].append(data_g[f"{tgt_lang}_system_score"])
    scores_dict[f"{tgt_lang}_system_score"].append(data_s[f"{tgt_lang}_system_score"])
```

```python
# comet
scores_dict = dict(scores_dict)
df = pd.DataFrame(scores_dict)
df = df.T
df.columns = ['googletrans', 'seamless']
df
```

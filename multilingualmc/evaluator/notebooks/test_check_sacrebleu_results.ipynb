```python
import pandas as pd
from multilingualmc.evaluator.eval_on_6060_gold import get_mean_std_dict
```

```python
dataset_path = '/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/eval/gold_predictions_dev_'
# dataset_path = '/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dataset/eval/'

df_list = []
for model in ['nllb', 'seamless', 'gpt4omini', 'aya', 'aya_old']:
    df1 = pd.read_csv(f"{dataset_path}{model}.csv")
    res_g = get_mean_std_dict(df1)
    df2 = pd.read_csv(f"{dataset_path}{model}_prompt_gpt4omini.csv")
    res_s = get_mean_std_dict(df2)
    df3 = pd.read_csv(f"{dataset_path}{model}_hard_replace.csv")
    res_h = get_mean_std_dict(df3)
    
    df4 = pd.read_csv(f"{dataset_path}{model}_hard_replace_morphology.csv")
    res_m = get_mean_std_dict(df4)
    
    df_g = pd.DataFrame.from_dict(res_g, orient='index')
    df_g.columns = [f'default_{model}']
    df_s = pd.DataFrame.from_dict(res_s, orient='index')
    df_s.columns = [f'prompt_{model}']
    df_h = pd.DataFrame.from_dict(res_h, orient='index')
    df_h.columns = [f'replace_{model}']
    
    df_m = pd.DataFrame.from_dict(res_m, orient='index')
    df_m.columns = [f'replace_morphology_{model}']
    
    df_list.extend([df_g, df_s, df_h, df_m])
df = pd.concat(df_list, axis=1)
pd.set_option('display.max_rows', None)
df
```

```python
# model = "nllb"
# model = "seamless"
# model = "gpt4omini"
# model = "aya_old"
# model = "aya"
for metric in ['BLEU', 'ChrF', 'ChrF++', 'TER']:
    print("\n" + metric)
    for model, model_name in [['nllb', 'nllb-200-3.3B'], ['seamless', 'hf-seamless-m4t-large'], ['gpt4omini', 'gpt-4o-mini'], ['aya_old', 'aya-23-8B'], ['aya', 'aya-expanse-8B']]:
        print(r"""\midrule
\multicolumn{4}{l}{\textit{\textbf{\texttt{""" + model_name + r"""}}}} \\
\specialrule{1.2pt}{0pt}{2.5pt}""")
        
        for tgt_lang in ['Arabic', 'Chinese', 'French', 'Japanese', 'Russian']:
            tmp = df[df.index.str.startswith(tgt_lang)]
            # Function to modify the indices as specified
            def modify_index(index):
                parts = index.split("_")
                new_index = " ".join(parts[1:])  # Remove the first element and join the rest with spaces
                return new_index.capitalize()   # Capitalize the first character

            tmp.index = tmp.index.map(modify_index)
            cols = [col for col in tmp.columns if model in col]
            # tmp.rename(columns = {f"default_{model}": "Direct Translation", f"prompt_{model}": "Prompting Refinement", f"replace_{model}": "Word Alignment"}, inplace=True)
            # print(tmp[["Direct Translation", "Prompting Refinement", "Word Alignment"]].to_latex(float_format="%.2f"))

            tmp.rename(
                columns={
                    f"default_{model}": "Direct Translation",
                    f"prompt_{model}": "Prompting Refinement",
                    f"replace_{model}": "Word Alignment",
                    f"replace_morphology_{model}": "Morphological Word Alignment",
                },
                index={
                    "Bleu score mean": "BLEU",
                    "Chrf score mean": "ChrF",
                    "Chrfpp score mean": "ChrF++",
                    "Ter score mean": "TER",
                },
                inplace=True,
            )
            print(
                tgt_lang + " & " + " & ".join(tmp[
                    [
                        "Direct Translation",
                        "Prompting Refinement",
                        "Word Alignment",
                        "Morphological Word Alignment",
                    ]
                ]
                .loc[[metric]]
                .iloc[0]
                .map(lambda x: f"{x:.2f}" if isinstance(x, float) else str(x))) + " \\\\"
                # .head(1)
                # .to_latex(float_format="%.2f", header=False)
                
            )
```

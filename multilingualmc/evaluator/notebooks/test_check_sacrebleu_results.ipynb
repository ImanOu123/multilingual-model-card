```python
import pandas as pd
from multilingualmc.evaluator.eval_on_6060_gold import get_mean_std_dict
```

```python
df_list = []
for model in ['nllb', 'seamless', 'gpt4omini']:
    df1 = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/eval/gold_predictions_dev_{model}.csv")
    res_g = get_mean_std_dict(df1)
    df2 = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/data_eval_6060/eval/gold_predictions_dev_{model}_prompt_gpt4omini.csv")
    res_s = get_mean_std_dict(df2)
    df_g = pd.DataFrame.from_dict(res_g, orient='index')
    df_g.columns = [f'default_{model}']
    df_s = pd.DataFrame.from_dict(res_s, orient='index')
    df_s.columns = [f'prompt_{model}']
    df_list.extend([df_g, df_s])
df = pd.concat(df_list, axis=1)
pd.set_option('display.max_rows', None)
df
```

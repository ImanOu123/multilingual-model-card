```python
import json

# Process the markdown content and convert it to a list of dictionaries
def convert_markdown_to_json(markdown_text):
    lines = markdown_text.strip().splitlines()
    result = []

    for line in lines:
        if "|" in line and not line.startswith("---"):
            # Extract the English and Chinese terms
            parts = line.split("|")
            if len(parts) >= 3:
                term_english = parts[1].strip()
                term_chinese = parts[2].strip()
                result.append({"term_english": term_english, "term_chinese": term_chinese})

    return result[1:]

# Convert the markdown content
terms_list = convert_markdown_to_json(markdown_content)

out_f = open("chinese.json", 'w')
# Convert the list of dictionaries to JSON
json.dump(terms_list, out_f, ensure_ascii=False, indent=2)
out_f.flush()
```

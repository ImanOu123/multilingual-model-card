```python
from bs4 import BeautifulSoup

# Function to extract terminologies and explanations from a local HTML file
def extract_terminologies_from_file(file_path):
    # Open and read the HTML file
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
    
    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(content, 'html.parser')
    
    # List to store dictionaries with the extracted data
    terminologies = []
    
    # Find all the divs containing the accordion items for English term and explanation
    accordion_items = soup.find_all('div', class_='elementor-accordion-item')

    # Find all Arabic term containers (elementor-widget-heading class)
    arabic_term_containers = soup.find_all('div', class_='elementor-widget-heading')[1:]
    print(len(accordion_items), len(arabic_term_containers))

    # Loop through each accordion item to extract terms and explanations
    for index, item in enumerate(accordion_items):
        # Extract the Arabic term: match it based on the index or nearby element
        if index < len(arabic_term_containers):
            arabic_term_container = arabic_term_containers[index]
            term_arabic = arabic_term_container.find(['h6'], class_='elementor-size-default')
            if term_arabic is None:
                continue
            else:
                term_arabic = term_arabic.get_text(strip=True)
        else:
            continue
            # term_arabic = "Unknown Term"

        # Extract the Arabic explanation (from the 'elementor-widget-text-editor' class)
        arabic_explanation_container = item.find_previous('div', class_='elementor-widget-text-editor')
        explanation_arabic = (
            arabic_explanation_container.get_text(strip=True) 
            if arabic_explanation_container else "No Arabic explanation found"
        )

        # Extract the English term and explanation
        english_term = item.find('a', class_='elementor-accordion-title').text.strip()
        english_explanation = item.find('div', class_='elementor-tab-content').text.strip()

        # Create a dictionary for the term
        term_dict = {
            "term_arabic": term_arabic,
            "term_english": english_term,
            "explanation_arabic": explanation_arabic,
            "explanation_english": english_explanation.split(":")[-1].strip()
        }

        # Add the dictionary to the list
        terminologies.append(term_dict)

    return terminologies

# Path to the downloaded HTML file
file_path = 'arabic.html'  # Replace with the actual path

# Extract the terminologies
arabic_terminologies = extract_terminologies_from_file(file_path)

# # Print the results
# for term in arabic_terminologies:
#     print(term)

import json
out_f = open("arabic.json", 'w')
json.dump(arabic_terminologies, out_f, ensure_ascii=False, indent=2)
out_f.flush()
```

```python
import pandas as pd
df_6060 = pd.read_csv("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/raw_data/6060_terminology_glossary.csv")
df_ref = pd.read_csv("/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/annotations/final_6060.csv")

tmp = set(df_6060['english'].tolist()) - set(df_ref['english'].tolist())


langs = ['Arabic', 'Chinese', 'French', 'Russian', 'Japanese']
for lang in langs:
    print(lang)
    # df = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/mturk/analysis/annotation_final_old/{lang}.csv")
    df = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/mturk/analysis/annotation_final/{lang}.csv")
    df.rename(columns={'word': 'English', "validated_translation": lang}, inplace=True)
    print(df.shape[0])
    df = df[~df['English'].isin(tmp)]
    print(df.shape[0])
    df[['English', lang]].to_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/mturk/analysis/annotation_final/{lang}.csv", index=True, index_label='Index')
```

```python
import nltk
from nltk.corpus import stopwords

# Download the NLTK stopwords if you haven't done so already
nltk.download('stopwords')

# Get the English stopwords from NLTK
stop_words = set(stopwords.words('english'))


for lang in langs:
    print(lang)
    df = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/mturk/analysis/annotation_final/{lang}.csv")['English'].tolist()
    # Find the overlap between the word list and stopwords
    
    overlapping_words = [word for word in df if word in stop_words]

    print("Overlapping Words:", overlapping_words)
```

```python
import pandas as pd

langs = ['Arabic', 'Chinese', 'French', 'Japanese', 'Russian']

def process_agreement_on_6060(df):
    df.rename(
        columns={'word': 'en_word', 'annotation6060': 'anno1_word', 'label6060': 'anno2_word'},
        inplace=True
    )
    
    columns = ["en_word", "anno1_word", "anno2_word"]
    df = df[columns]
    df['source'] = '6060'
    return df

def process_gpt4o_vs_threshold(df):
    df.rename(
        columns={'word': 'en_word', '1)validated_translation_by_gpt4o': 'anno1_word', '2)word>=0.5': 'anno2_word'},
        inplace=True
    )
    
    columns = ["en_word", "anno1_word", "anno2_word"]
    df = df[columns]
    df['source'] = 'threshold'
    return df

def reformat_df(df):
    n = 5
    columns = ["en_word", "anno1_word", "anno2_word", 'source']
    padding_rows = n - (len(df) % n) if len(df) % n != 0 else 0
    if padding_rows > 0:
        padding = pd.DataFrame({col: ["-"] * padding_rows for col in columns})
        df = pd.concat([df, padding], ignore_index=True)

    # Convert to a list of groups, flatten each group, and reconstruct the DataFrame
    grouped_data = [
        df.iloc[i:i + n].values.flatten() for i in range(0, len(df), n)
    ]

    # Create new column names dynamically
    new_columns = [f"{col}{j+1}" for j in range(n) for col in columns]

    # Create the final DataFrame
    reshaped_df = pd.DataFrame(grouped_data, columns=new_columns)
    return reshaped_df


for lang in langs:
    df1 = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/mturk/analysis/annotation_results_crawled/check_agreement_on_6060/{lang}.csv")
    df1 = process_agreement_on_6060(df1)
    
    df2 = pd.read_csv(f"/home/jiaruil5/multilingual/multilingual-model-card/multilingualmc/dictionary_collection/mturk/analysis/annotation_results_crawled/check>0.5new/{lang}.csv")
    df2 = process_gpt4o_vs_threshold(df2)
    
    df = pd.concat([df1, df2], ignore_index=True)
    df = reformat_df(df)
    df.to_csv(f"dataset/{lang}.csv", index=False)
```

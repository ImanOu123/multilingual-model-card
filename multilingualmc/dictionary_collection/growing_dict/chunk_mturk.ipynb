```python
import json
import numpy as np
import pandas as pd
data = json.load(open("mturk.json", 'r'))
df = pd.DataFrame().from_records(data)
df
```

```python
# Creating interleaved sub-dataframes from the original DataFrame.
# Split the DataFrame into 5 sub-dataframes interleavingly.

# Create empty sub-dataframes
interleaved_sub_dfs = [df.iloc[i::5].reset_index(drop=True) for i in range(5)]

# Add suffixes to each interleaved sub-dataframe
suffixes = ['1', '2', '3', '4', '5']
interleaved_sub_dfs = [sub_df.add_suffix(suffix) for sub_df, suffix in zip(interleaved_sub_dfs, suffixes)]

# Concatenate the interleaved sub-dataframes by columns
interleaved_concat_df = pd.concat(interleaved_sub_dfs, axis=1)

interleaved_concat_df
```

```python
interleaved_concat_df.to_csv("mturk_5terms_per_page.csv")
```

```python
interleaved_concat_df.sample(5).reset_index(drop=True).to_csv("test_5terms_per_page.csv")
```
